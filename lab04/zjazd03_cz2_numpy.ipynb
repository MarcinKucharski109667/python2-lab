```python
import numpy
```

```python
# Ceny 4 produktów w 3 sklepach
ceny = np.array([[29.99, 31.99, 28.99],    # Produkt A
                  [49.99, 52.99, 47.99],    # Produkt B
                  [14.99, 15.99, 13.99],    # Produkt C
                  [99.99, 104.99, 94.99]])  # Produkt D

sklepy = ['Centrum', 'Galeria', 'Online']
produkty = ['A', 'B', 'C', 'D']
print(f"Kształt macierzy: {ceny.shape}")  # (4, 3)
```

```python
ceny_inflacja = ceny * 1.05

print("Ceny po inflacji 5%:")
print(ceny_inflacja)
```

```python
rabaty = numpy.array([0.10, 0.05, 0.15])
ceny_po_rabacie = ceny * (1 - rabaty)

print("Ceny z uwzględnieniem rabatów:")
print(ceny_po_rabacie)
```

```python
marza = numpy.array([[0.30], [0.40], [0.25], [0.50]])
koszt_zakupu = ceny / (1 + marza)

print("Koszt zakupu każdego produktu:")
print(koszt_zakupu)
```

```python
najtansze_indeksy = numpy.argmin(ceny, axis=1)

print("Najtańszy sklep dla poszczególnych produktów:")
for i in range(len(produkty)):
    najtanszy_sklep = sklepy[najtansze_indeksy[i]]
    print(f"Produkt {produkty[i]}: {najtanszy_sklep}")
```

```python
# Dane miesięczne z jednego roku (flat)
dane_flat = np.array([120, 135, 98, 142, 167, 189, 201, 178, 156, 143, 198, 221])

# Przekształć na kwartały (4 kwartały × 3 miesiące)
kwartaly = dane_flat.reshape(4, 3)
print(f"Kwartały:\n{kwartaly}")
print(f"Suma per kwartał: {kwartaly.sum(axis=1)}")
print(f"Najlepszy kwartał: Q{kwartaly.sum(axis=1).argmax() + 1}")
```

```python
# Masz dane o sprzedaży 6 produktów jako wektor
sprzedaz_flat = np.array([100, 200, 150, 300, 250, 180, 120, 220, 170, 310, 260, 190])
# Wiesz, że to 6 produktów × 2 miesiące

# 1. Reshape na macierz 6×2
sprzedaz_macierz = sprzedaz_flat.reshape(-1, 2)

print("Macierz 6x2 (sprzedaż 6 produktów w 2 miesiącach):")
print(sprzedaz_macierz)

# 2. Oblicz średnią sprzedaż per produkt (axis=1)
srednia_per_produkt = sprzedaz_macierz.mean(axis=1)
print("\nŚrednia sprzedaż dla każdego produktu:")
print(srednia_per_produkt)
# 3. Oblicz zmianę między miesiącem 1 a 2 (kolumna 1 - kolumna 0)
zmiana_sprzedazy = sprzedaz_macierz[:, 1] - sprzedaz_macierz[:, 0]
print("\nZmiana sprzedaży (Miesiąc 2 - Miesiąc 1):")
print(zmiana_sprzedazy)
```

```python
np.random.seed(2026)

# 8 pracowników, dane roczne
pracownicy = ['Anna', 'Jan', 'Ewa', 'Marek', 'Kasia', 'Piotr', 'Zofia', 'Tomek']
pensje = np.array([5500, 7200, 4800, 9100, 6300, 4200, 8500, 5800])
staz = np.array([3, 8, 2, 15, 5, 1, 12, 4])
oceny = np.array([4.2, 3.8, 4.5, 4.0, 4.7, 3.5, 4.3, 3.9])

# Sprzedaż miesięczna 4 produktów przez 6 miesięcy
produkty_nazwy = ['Laptop', 'Tablet', 'Smartfon', 'Akcesorium']
sprzedaz = np.random.randint(20, 200, size=(4, 6))
ceny_produktow = np.array([3500, 1800, 2500, 150])
```

```python
# 1. Oblicz średnią, medianę i odchylenie standardowe pensji
srednia = pensje.mean()
mediana = numpy.median(pensje)
odchylenie = pensje.std()

print(f"Średnia pensja: {srednia:.2f}")
print(f"Mediana pensji: {mediana:.2f}")
print(f"Odchylenie standardowe: {odchylenie:.2f}")

# 2. Wyświetl imiona osób zarabiających powyżej średniej
pracownicy_arr = numpy.array(pracownicy)
najlepiej_zarabiajacy = pracownicy_arr[pensje > srednia]

print("Osoby zarabiające powyżej średniej:")
print(najlepiej_zarabiajacy)()]

# 3. Wyświetl imiona osób zarabiających poniżej mediany
najgorzej_zarabiajacy = pracownicy_arr[pensje < mediana]

print("Osoby zarabiające poniżej mediany:")
print(najgorzej_zarabiajacy)
# 4. Oblicz rozstęp (max - min) i rozstęp międzykwartylowy (IQR)
rozstep = pensje.max() - pensje.min()
iqr = numpy.percentile(pensje, 75) - numpy.percentile(pensje, 25)

print(f"Rozstęp pensji (max - min): {rozstep:.2f}")
print(f"Rozstęp międzykwartylowy (IQR): {iqr:.2f}")
```

```python
# Zasady premii:
# - ocena >= 4.5 → premia 2000 zł
# - ocena >= 4.0 → premia 1000 zł
# - ocena < 4.0 → premia 0 zł
# Podpowiedź: zagnieżdżony where:
premia = numpy.where(oceny >= 4.5, 2000, numpy.where(oceny >= 4.0, 1000, 0))

print("Wyliczone premie:")
print(premia)

nowe_wynagrodzenie = pensje + premia

print("Nowe wynagrodzenie (pensja + premia):")
print(nowe_wynagrodzenie)
# Oblicz nowe wynagrodzenie (pensja + premia)
indeks_max_premii = numpy.argmax(premia)
najlepszy_pracownik = pracownicy[indeks_max_premii]

print(f"Najwyższą premię otrzymał(a): {najlepszy_pracownik}")
```

```python
# Posortuj pracowników wg pensji (od najwyższej)
# 1. Uzyskanie indeksów posortowanych malejąco według pensji
kolejnosc = numpy.argsort(pensje)[::-1]

print("Ranking pracowników według pensji:")
print("-" * 40)

# Wyświetl ranking: pozycja, imię, pensja, staż
for i, idx in enumerate(kolejnosc, 1):
    imie = pracownicy[idx]
    pensja_pracownika = pensje[idx]
    staz_pracownika = staz[idx]
    
    print(f"{i}. {imie} | Pensja: {pensja_pracownika} zł | Staż: {staz_pracownika} lat(a)")
```

```python
# Oblicz korelację: staż vs pensja
korelacja_staz_pensja = numpy.corrcoef(staz, pensje)[0, 1]
korelacja_staz_ocena = numpy.corrcoef(staz, oceny)[0, 1]

print(f"Korelacja (staż vs pensja): {korelacja_staz_pensja:.4f}")
print(f"Korelacja (staż vs ocena): {korelacja_staz_ocena:.4f}")

# Oblicz korelację: staż vs ocena
sila_staz_pensja = numpy.abs(korelacja_staz_pensja)
sila_staz_ocena = numpy.abs(korelacja_staz_ocena)

print("-" * 40)
if sila_staz_pensja > sila_staz_ocena:
    print("Związek między stażem a pensją jest silniejszy.")
elif sila_staz_ocena > sila_staz_pensja:
    print("Związek między stażem a oceną pracownika jest silniejszy.")
else:
    print("Oba związki mają identyczną siłę.")
```

```python
# Oblicz przychód: sprzedaz (4×6) * ceny_produktow (4,) — UWAGA na broadcasting!
# Problem: (4,6) * (4,) — nie pasuje! 6 != 4
# Rozwiązanie: ceny muszą być kolumną (4×1)
ceny_kol = ceny_produktow.reshape(4, 1)
przychod = sprzedaz * ceny_kol

print("Macierz przychodów:")
print(przychod)

# Łączny przychód dla poszczególnych produktów (suma w wierszach)
przychod_per_produkt = przychod.sum(axis=1)
najlepszy_produkt_idx = numpy.argmax(przychod_per_produkt)

# Łączny przychód dla poszczególnych miesięcy (suma w kolumnach)
przychod_per_miesiac = przychod.sum(axis=0)
najlepszy_miesiac_idx = numpy.argmax(przychod_per_miesiac)

print(f"Największy łączny przychód przyniósł produkt o indeksie: {najlepszy_produkt_idx}")
print(f"Najlepszy miesiąc łącznie to miesiąc o indeksie: {najlepszy_miesiac_idx}")

# Sumowanie odpowiednich wycinków macierzy
suma_pierwsze_3 = przychod[:, :3].sum()
suma_ostatnie_3 = przychod[:, 3:].sum()

print(f"Suma przychodów z pierwszych 3 miesięcy: {suma_pierwsze_3}")
print(f"Suma przychodów z ostatnich 3 miesięcy: {suma_ostatnie_3}")

print("-" * 40)
# Porównanie trendu
if suma_ostatnie_3 > suma_pierwsze_3:
    print("Trend jest rosnący – ostatnie 3 miesiące były lepsze.")
elif suma_ostatnie_3 < suma_pierwsze_3:
    print("Trend jest malejący – pierwsze 3 miesiące były lepsze.")
else:
    print("Przychody w obu okresach były na dokładnie takim samym poziomie.")
```
